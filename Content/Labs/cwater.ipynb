# Bepaling van soortelijke warmte van water


# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

De soortelijke warmte van water is de hoeveelheid warmte die nodig is om een kilogram aan water 1 graad Celsius of Kelvin te verhogen en is te berekenen met $$c =\frac{Q}{m\Delta T} $$. In het practicum wordt er energie toegevoegd met behulp van en voedingsbron. De energie die toegevoegd wordt staat gelijk aan $$ Q = P \cdot t$$.
Waarbij $$ P = U \cdot I$$.

# Resultaten



````python
# Hier de data en de analyse
# Metingen waren alleen gedaan met 15 V en 1.5 A vanwege wat wel en niet mocht van de T.A. voor de veiligheid en behoud van de apparatuur
import numpy as np
U = 15
I= 1.5
Massa_beker = 0.2824
Massa_beker_water = 0.8807
massa_water = Massa_beker_water - Massa_beker
Begintemperatuur = 19.5 
t = np.array([101, 219, 329, 440, 588])
P = U*I
Q = P*t
delta_T = np.array([1,2,3,4,5])
c_ =  np.divide(Q,massa_water*delta_T)
c= np.average(c_)
print('De soortelijke warmte van water is ',round(c),'J/kgK')
````

# Discussie en conclusie

Hier een korte discussie en conclusie over de resultaten van het experiment en de implicaties daarvan.


Uit dit practicum is de gevonden waarde voor de soortelijke warmte van water 4120 J/kgK. Dit ligt dicht bij de theoretische waarde van 4186 J/kgK, dus het experiment was redelijk succesvol.
Het tijdverschil tussen opeenvolgende temperatuurstijgingen hoort volgens de theorie constant te zijn (Δt zou gelijk moeten zijn voor iedere graad stijging), maar uit de resultaten blijkt dat dit niet het geval is. De tijd neemt toe naarmate de temperatuur stijgt.
Een waarschijnlijke oorzaak hiervan is dat het warmteverlies naar de omgeving groter wordt naarmate het temperatuurverschil tussen het water en de omgeving toeneemt. Dit volgt uit de wet van Newton voor afkoeling:
$$Q_{verlies} = h \cdot A \cdot \Delta T_{omgeving}$$

Omdat de temperatuur van het water stijgt, wordt ΔT groter, dus groter warmteverlies. Dit betekent dat niet alle elektrische energie in het water gaat, maar een deel verloren gaat aan de omgeving.Voor een verbeterd onderzoek plaats het waterbad in een geïsoleerde container en Herhaal het experiment meerdere keren.
