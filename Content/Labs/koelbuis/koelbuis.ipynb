# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur.

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen.

# Resultaten

`````python
# Hier de data en de analyse

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def exp_func(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

dikte_wand = 0.00115 # m
radius = 0.038 / 2
hoogte = 0.0915 
buitenoppervlak = 2*radius*np.pi*hoogte # bepaal zelf in m^2
print(f'Buitenoppervlak: {buitenoppervlak} m^2')
omgevingstemperatuur = 298.25 # bepaal de omgevingstemperatuur in graden C
print(f'Omgevingstemperatuur: {omgevingstemperatuur} K')
warmtecapaciteit = 3.8e2 # bepaal de warmtecapaciteit in J/K
print(f'Warmtecapaciteit: {warmtecapaciteit} J/K')

met_dop = np.genfromtxt('koelbuis_met_dop.csv', delimiter=';', skip_header=1)
zonder_dop = np.genfromtxt('koelbuis_zonder_dop.csv', delimiter=';', skip_header=1)

times_met_dop = met_dop[:,0]  # tijd in seconden
temps_met_dop = met_dop[:,1]  # temperatuur in graden C
temps_met_dop = temps_met_dop + 273.15  # omzetten naar Kelvin

times_zonder_dop = zonder_dop[:,0]  # tijd in seconden
temps_zonder_dop = zonder_dop[:,1]  # temperatuur in graden C
temps_zonder_dop = temps_zonder_dop + 273.15  # omzetten naar Kelvin

# pas beginwaardes aan naar schatting
popt, pocv = curve_fit(exp_func, times_met_dop, temps_met_dop, p0=[30, 100, omgevingstemperatuur], maxfev=5000)

A_exp, tau_exp, T_omg_exp = popt

y_fit = exp_func(times_met_dop, *popt)

popt_zonder, pocv_zonder = curve_fit(exp_func, times_zonder_dop, temps_zonder_dop, p0=[30, 100, omgevingstemperatuur], maxfev=5000)
A_exp_zonder, tau_exp_zonder, T_omg_exp_zonder = popt_zonder
y_fit_zonder = exp_func(times_zonder_dop, *popt_zonder)

plt.figure()
plt.xlabel('t (s)')
plt.ylabel('T (K)')

plt.plot(times_met_dop, temps_met_dop, 'bo', label='Meting met dop')
plt.plot(times_met_dop, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg_exp))
plt.plot(times_zonder_dop, temps_zonder_dop, 'go', label='Meting zonder dop')
plt.plot(times_zonder_dop, y_fit_zonder, 'm-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp_zonder, tau_exp_zonder, T_omg_exp_zonder))

plt.legend()

plt.show()

h_exp = (warmtecapaciteit) / (tau_exp * buitenoppervlak)
h_exp_zonder = (warmtecapaciteit) / (tau_exp_zonder * buitenoppervlak)
 
print(f'Warmteoverdrachtcoefficient met dop: {h_exp} W/m^2 K') # warmteoverdrachtscoëfficiënt in W/m^2 K
print(f'Warmteoverdrachtcoefficient zonder dop: {h_exp_zonder} W/m^2 K') # warmteoverdrachtscoëfficiënt in W/m^2 K
# Sla figuren op met  
# 
plt.savefig("fit.png", dpi=450)
`````

# Discussie en conclusie
Voor koelbuis met dop is $\tau = 202.9$ groter dan zonder dop $\tau = 142.2$, wat betekent dat de koelbuis met dop er langer over doet om af te koelen. Door weghalen van de dop stijgt de warmteoverdrachtcoefficient h met $73.2 W/m^3K$. Dit komt doordat de lucht na het weghalen van de dop door de buis kan stromen, waardoor de effecitieve oppervlakte voor warmteoverdracht groter wordt en is er meer convectie.
We verwachten dat convecttie het grootst deel van de warmtetransport uitmaakt, omdat door het weghalen van de buis h flink stijgt. De contributie van straling zou stuk kleiner zijn omdat de temperatuur slechts maximum 43 graden celcius is. Geleiding zou het kleinst aandeel hebben omdat de warmte slechts via lucht en de punt waar de buis de houder raank. Lucht is een slechte geleider en de raakvlak is zeer klein.
